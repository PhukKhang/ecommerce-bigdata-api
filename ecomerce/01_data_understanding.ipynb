```python
def load_csv(path):
    return (
        spark.read
        .option("header", True)
        .option("inferSchema", True)
        .option("multiLine", True)
        .option("quote", '"')
        .option("escape", '"')
        .csv(path)
    )
```

Tìm kiếm dữ liệu null

```python
from pyspark.sql.functions import *

def check_null(df):
    return df.select([
        count(
            when(col(c).isNull(), c)
        ).alias(c)
        for c in df.columns
    ])
```

```python
# NULL CHECK
from pyspark.sql.functions import col, when, count

tables = {
    "customers": customers_df,
    "orders": orders_df,
    "products": products_df,
    "sellers": sellers_df,
    "reviews": reviews_df,
    "payments": payments_df,
    "order_items": items_df,
    "category_translation": category_df
}

for table_name, df in tables.items():

    print("\n" + "=" * 100)
    print(f"TABLE: {table_name}")
    print("=" * 100)

    total_rows = df.count()

    null_df = df.select([
        count(
            when(col(c).isNull(), c)
        ).alias(c)
        for c in df.columns
    ])

    null_row = null_df.collect()[0].asDict()

    found_null = False

    for column_name, null_count in null_row.items():

        if null_count > 0:

            found_null = True

            null_percent = (
                null_count / total_rows * 100
            )

            print(
                f"{column_name:<45}"
                f"NULL = {null_count:<10}"
                f"({null_percent:.2f}%)"
            )

    if not found_null:
        print("No NULL values found.")
```

Kiểm tra Duplicate


```python
# CHECK DUPLICATE RECORDS

tables = {
    "customers": customers_df,
    "orders": orders_df,
    "products": products_df,
    "sellers": sellers_df,
    "reviews": reviews_df,
    "payments": payments_df,
    "order_items": items_df,
    "category_translation": category_df
}

print("=" * 80)
print("DUPLICATE CHECK REPORT")
print("=" * 80)

for table_name, df in tables.items():

    total_rows = df.count()
    unique_rows = df.dropDuplicates().count()
    duplicate_rows = total_rows - unique_rows

    duplicate_percent = (
        duplicate_rows / total_rows * 100
        if total_rows > 0 else 0
    )

    print(f"\nTable: {table_name}")
    print(f"Total Rows      : {total_rows:,}")
    print(f"Unique Rows     : {unique_rows:,}")
    print(f"Duplicate Rows  : {duplicate_rows:,}")
    print(f"Duplicate Rate  : {duplicate_percent:.4f}%")

    if duplicate_rows == 0:
        print("Status          : No duplicates")
    else:
        print("Status          : Contains duplicates")

print("\n" + "=" * 80)
print("CHECK COMPLETED")
print("=" * 80)
```

```python
# CHECK DUPLICATE PRIMARY KEY

from pyspark.sql.functions import col, count

pk_mapping = {
    "customers": ("customer_id", customers_df),
    "orders": ("order_id", orders_df),
    "products": ("product_id", products_df),
    "sellers": ("seller_id", sellers_df),
    "reviews": ("review_id", reviews_df)
}

print("=" * 80)
print("PRIMARY KEY DUPLICATE CHECK")
print("=" * 80)

for table_name, (pk, df) in pk_mapping.items():

    dup_count = (
        df.groupBy(pk)
          .count()
          .filter(col("count") > 1)
          .count()
    )

    print(f"\nTable: {table_name}")
    print(f"Primary Key: {pk}")
    print(f"Duplicated Keys: {dup_count}")

    if dup_count == 0:
        print("Status: VALID")
    else:
        print("Status: INVALID")

print("\nValidation completed.")
```
